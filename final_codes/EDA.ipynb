```python
dataset = pd.read_csv('../Data/Product_Normalization_GRI_Expanded.csv')
dataset.head()
```

```python
sampled_descriptions = dataset.groupby('Guest Room Info').apply(
    lambda x: x['Room Description Expanded'].sample()
)

print(sampled_descriptions)
```

```python
#%%
# Count occurrences of ' bed ' (case insensitive, with word boundaries)
bed_counts = dataset['Room Description Expanded'].str.count(r'\b bed \b', flags=re.IGNORECASE)

print(f"Total occurrences of 'bed': {bed_counts.sum()}")
print(f"Total Descriptions: {len(dataset)}")
```

```python
#%%
# Count occurrences of ' bed room ' (case insensitive, with word boundaries)
bed_room_counts = dataset['Room Description Expanded'].str.count(r'\b bed room \b', flags=re.IGNORECASE)

print(f"Total occurrences of ' bed room ': {bed_room_counts.sum()}")
print(f"Total Descriptions: {len(dataset)}")
```
