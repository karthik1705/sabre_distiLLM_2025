```python

########################################
# 1) READ CSV FILES
########################################
DESCRIPTIONS_CSV = "descriptions.csv"
ATTRIBUTES_CSV = "attributes.csv"

def read_csv_data(descriptions_path, attributes_path):
    """
    Reads two CSVs:
      - descriptions.csv: columns = [description_id, text]
      - attributes.csv: columns = [description_id, room_type, bed_type, room_amenities (maybe comma-separated?), ...]
    Returns a list of dicts: 
      [
        {
          'description_id': ...,
          'text': '...',
          'room_type': 'Standard',
          'bed_type': 'Queen',
          'amenities': ['Accessible', 'Non-Smoking'],
          ...
        },
        ...
      ]
    """
    # Read descriptions
    desc_map = {}
    with open(descriptions_path, "r", encoding="utf-8") as f:
        reader = csv.DictReader(f)
        for row in reader:
            d_id = row["description_id"]
            desc_map[d_id] = row["text"]
    
    # Read attributes
    data = []
    with open(attributes_path, "r", encoding="utf-8") as f:
        reader = csv.DictReader(f)
        for row in reader:
            d_id = row["description_id"]
            record = {
                "description_id": d_id,
                "text": desc_map[d_id],
                "room_type": row["room_type"].strip(),  # e.g. "Deluxe"
                "bed_type": row["bed_type"].strip(),    # e.g. "Queen"
            }
            # Suppose 'room_amenities' is comma-separated
            # e.g. "Accessible,Non-Smoking" => ["Accessible", "Non-Smoking"]
            amenities = row["room_amenities"].split(",")
            record["amenities"] = [a.strip() for a in amenities if a.strip()]
            
            data.append(record)
    return data

raw_data = read_csv_data(DESCRIPTIONS_CSV, ATTRIBUTES_CSV)
```

```python
########################################
# 2) BUILD A MULTI-LABEL SCHEME
########################################
# Let's say you have known sets for each:
ALL_ROOM_TYPES = room_types_stripped_list
ALL_BED_TYPES = bed_types_stripped_list
ALL_AMENITIES = room_amenities_stripped_list

# We'll flatten them into one multi-label vector, e.g.:
#   [ Standard, Deluxe, Suite, Executive,
#     King, Queen, Double,
#     Accessible, Non-Smoking, Ocean-View, Balcony, Kitchen ]
# For each record, exactly ONE room type is 1, EXACTLY ONE bed type is 1, 
# and zero or more amenities are 1.

label_list = ALL_ROOM_TYPES + ALL_BED_TYPES + ALL_AMENITIES
label2id = {lab: i for i, lab in enumerate(label_list)}
id2label = {i: lab for lab, i in label2id.items()}

def build_multilabel_vector(room_type, bed_type, amenities):
    vec = [0]*len(label_list)
    if room_type in label2id:
        vec[label2id[room_type]] = 1
    if bed_type in label2id:
        vec[label2id[bed_type]] = 1
    for a in amenities:
        if a in label2id:
            vec[label2id[a]] = 1
    return vec
```

```python
import os
from tokenizers import ByteLevelBPETokenizer

########################################
# Step 1: Gather Text Data
########################################
# Suppose you have 35K hotel room descriptions in a list `all_texts`.
# If they're in a CSV, just read them into a list first.

# Example toy data
all_texts = descriptions

########################################
# Step 2: Write Texts to a File for BPE Training
########################################
with open("corpus.txt", "w", encoding="utf-8") as f:
    for line in all_texts:
        # Basic cleaning if you want
        f.write(line.strip() + "\n")

########################################
# Step 3: Train a BPE Tokenizer
########################################
# We'll train a Byte-Level BPE tokenizer with a small vocab_size for illustration
bpe_tokenizer = ByteLevelBPETokenizer()

bpe_tokenizer.train(
    files=["corpus.txt"],
    vocab_size=10000,         # adjust as needed
    min_frequency=1,        # can be 2 or higher if you want to filter rare merges
    special_tokens=["<s>", "</s>", "<unk>", "<pad>", "<mask>"]
)

# This creates merges.txt and vocab.json in the current dir
bpe_tokenizer.save_model("bpe_artifacts")

# Now we have a custom BPE tokenizer that knows how to split these hotel descriptions.

########################################
# Step 4: Define a Simple Phrase-Labeling Heuristic
########################################
# We'll define a function that:
#  1) Takes text
#  2) Tokenizes it using our BPE tokenizer
#  3) Produces an IOB tag sequence: "B-PHRASE", "I-PHRASE", or "O"
#
# For demonstration, we label a token as part of a phrase if it's "long enough."
# The first token in a run is B-PHRASE, subsequent tokens are I-PHRASE.

def label_subword_phrases_bpe(text, tokenizer, length_threshold=6):
    """
    1) Encode text using BPE
    2) Return (tokens, iob_tags), 
       where iob_tags is "B-PHRASE", "I-PHRASE", or "O"
    """
    encoding = tokenizer.encode(text)
    bpe_tokens = encoding.tokens  # e.g. ["This", "del", "uxe", "room", ...] depending on merges

    iob_tags = []
    in_phrase = False
    for i, tok in enumerate(bpe_tokens):
        # Byte-level BPE might add special chars like "Ġ". Let's strip them out for length checking:
        clean_tok = tok.replace("Ġ", "").replace("##", "")

        if len(clean_tok) >= length_threshold:
            # We'll say this is part of a "phrase"
            if not in_phrase:
                iob_tags.append("B-PHRASE")
                in_phrase = True
            else:
                iob_tags.append("I-PHRASE")
        else:
            iob_tags.append("O")
            in_phrase = False

    return bpe_tokens, iob_tags
```

```python
########################################
# Step 5: Apply Phrase Extraction to Each Description
########################################
processed = []
for text in all_texts:
    bpe_toks, iob_tags = label_subword_phrases_bpe(text, bpe_tokenizer)
    processed.append({
        "text": text,
        "bpe_tokens": bpe_toks,
        "iob_tags": iob_tags
    })

########################################
# Step 6: Inspect or Export
########################################
# Let's just print them out for manual inspection:
for item in processed:
    print("Original Text:", item["text"])
    print("BPE Tokens:   ", item["bpe_tokens"])
    print("IOB Tags:     ", item["iob_tags"])
    print("----")
```

```python
import re
from collections import Counter
import numpy as np

all_texts = descriptions

# 1) Extract sub-phrase tokens for each doc
docs_tokens = []
for text in all_texts:
    bpe_toks, iob_tags = label_subword_phrases_bpe(text, bpe_tokenizer)
    # keep only the tokens labeled B-PHRASE or I-PHRASE
    phrase_toks = []
    for tok, tag in zip(bpe_toks, iob_tags):
        if tag in ("B-PHRASE", "I-PHRASE"):
            # remove special chars like 'Ġ'
            phrase_toks.append(tok.replace("Ġ", "").replace("##", ""))
    docs_tokens.append(phrase_toks)

# 2) Build a global vocabulary of sub-phrase tokens
#    e.g. for all docs, count how many times each token appears
global_counter = Counter()
for tok_list in docs_tokens:
    global_counter.update(tok_list)

# Optionally, filter out tokens that appear <5 times, etc.
min_count = 2
vocab = [tok for tok, cnt in global_counter.items() if cnt >= min_count]
vocab2id = {v: i for i, v in enumerate(vocab)}

# 3) Create a bag-of-phrases vector for each doc
#    We'll do a simple count vector. You could do TF-IDF, etc.
doc_vectors = []
for tok_list in docs_tokens:
    vec = np.zeros(len(vocab), dtype=np.float32)
    for tok in tok_list:
        if tok in vocab2id:
            vec[vocab2id[tok]] += 1
    doc_vectors.append(vec)

doc_vectors = np.array(doc_vectors)  # shape [num_docs, vocab_size]
```

```python
# Simple TF-IDF transform
from sklearn.feature_extraction.text import TfidfTransformer
transformer = TfidfTransformer()
doc_vectors_tfidf = transformer.fit_transform(doc_vectors)
```

```python
from sklearn.cluster import KMeans

k = 35  # example
kmeans = KMeans(n_clusters=k, random_state=42)
kmeans.fit(doc_vectors_tfidf)

cluster_assignments = kmeans.labels_  # each doc's cluster id
```

```python
def analyze_cluster(cluster_id, results_df, top_n=10):
    """
    Analyzes a specific cluster from results_with_labels
    
    Args:
        cluster_id: ID of cluster to analyze
        results_df: DataFrame with 'cluster_id', 'Room Description' columns
        top_n: Number of top terms to return
    """
    # Get descriptions in this cluster
    cluster_texts = results_df[results_df['cluster_id'] == cluster_id]['Room Description Expanded']
    
    # Create TF-IDF for just this cluster
    cluster_vectorizer = TfidfVectorizer(stop_words='english')
    cluster_tfidf = cluster_vectorizer.fit_transform(cluster_texts)
    
    # Get most common/important terms
    feature_names = cluster_vectorizer.get_feature_names_out()
    scores = cluster_tfidf.sum(axis=0).A1
    top_terms = sorted(zip(feature_names, scores), key=lambda x: x[1], reverse=True)[:top_n]
    
    return {
        'size': len(cluster_texts),
        'top_terms': top_terms,
        'sample_descriptions': cluster_texts.head().tolist()
    }

def suggest_room_type(cluster_analysis, room_types_list):
    """Suggests room type based on cluster's top terms"""
    room_type_matches = {}
    
    for room_type in room_types_list:
        room_words = set(room_type.lower().split())
        score = sum(1 for term, _ in cluster_analysis['top_terms'] 
                   if any(word in term.lower() for word in room_words))
        room_type_matches[room_type] = score
    
    return max(room_type_matches.items(), key=lambda x: x[1])[0]
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
def custom_tfidf_clustering(descriptions, room_types_list, n_clusters=35, boost_factor=2.0):
    """
    Enhanced TF-IDF clustering with room type term boosting
    """
    # 1. Create room type vocabulary
    room_type_words = set()
    for label in room_types_list:
        room_type_words.update(label.lower().split())
    
    # 2. Create custom TF-IDF
    vectorizer = TfidfVectorizer(
        lowercase=True,
        stop_words='english',
        max_features=10000
    )
    
    # Get base TF-IDF
    tfidf_matrix = vectorizer.fit_transform(descriptions)
    feature_names = vectorizer.get_feature_names_out()
    
    # 3. Boost room type related terms
    boost_indices = [
        idx for idx, term in enumerate(feature_names)
        if any(room_word in term for room_word in room_type_words)
    ]
    
    # Modify TF-IDF matrix
    boosted_matrix = tfidf_matrix.copy()
    boosted_matrix[:, boost_indices] *= boost_factor
    
    # 4. Cluster with boosted features
    kmeans = KMeans(n_clusters=n_clusters, random_state=42)
    clusters = kmeans.fit_predict(boosted_matrix)
    
    # 5. Create results
    results_df = pd.DataFrame({
        'Room Description': descriptions,
        'cluster_id': clusters
    })
    
    return results_df, kmeans, vectorizer
```

```python

########################################
# 4) TOKENIZE WITH A PRETRAINED MODEL
########################################
model_name = "distilbert-base-uncased"  # smaller, faster on CPU
tokenizer = AutoTokenizer.from_pretrained(model_name)

# For token classification, we need an alignment from our naive tokens -> subword tokens
# We'll do a simple alignment approach.

def tokenize_and_align_labels(examples):
    # We'll join the naive tokens with space, then let the HF tokenizer handle subwords
    joined_text = [" ".join(toks) for toks in examples["tokens"]]
    tokenized = tokenizer(
        joined_text,
        padding="max_length",
        truncation=True,
        max_length=128,
        return_tensors="pt",
        return_offsets_mapping=True,
    )
    
    all_labels_token = []
    for i, offsets in enumerate(tokenized["offset_mapping"]):
        naive_tokens = examples["tokens"][i]
        naive_iob = examples["subword_iob"][i]
        
        # We'll create a label array for each subword in the tokenized output
        word_ids = []
        # A simpler approach: the HF tokenizers usually have a "word_ids()" method 
        # if we pass "is_split_into_words=True", but we didn't do that here. 
        # So let's do a manual approach:
        # We'll match the naive token boundaries to offset ranges.
        
        subword_labels = []
        current_naive_idx = 0
        naive_token_starts = []
        cum_length = 0
        
        # Build naive token boundary offsets for alignment
        # e.g., token "hello" => length=5, next offset starts at 5 (plus a space if you want)
        naive_offsets = []
        char_pos = 0
        for tok_idx, tok_str in enumerate(naive_tokens):
            start_pos = char_pos
            end_pos = char_pos + len(tok_str)
            naive_offsets.append((start_pos, end_pos))
            char_pos = end_pos + 1  # plus space
        
        for (sub_start, sub_end) in offsets.tolist():
            if sub_start == 0 and sub_end == 0:
                # This is a special token (CLS, SEP, PAD), label as "O" or -100 for ignoring
                subword_labels.append("O")
                continue
            
            # Find which naive token this subword belongs to
            # We do a naive approach: if sub_start,sub_end is within naive_offsets
            # Because we joined tokens with spaces, let's see if we can find it quickly:
            matched_idx = None
            for idx, (n_start, n_end) in enumerate(naive_offsets):
                # subword offset within n_start..n_end => belongs to naive token idx
                if sub_start >= n_start and sub_end <= n_end:
                    matched_idx = idx
                    break
            if matched_idx is not None:
                subword_labels.append(naive_iob[matched_idx])
            else:
                # no match => "O"
                subword_labels.append("O")
        
        all_labels_token.append(subword_labels)
    
    tokenized["labels"] = examples["labels"]  # multi-label vector
    tokenized["token_iob"] = all_labels_token
    return tokenized

dataset_tokenized = dataset_dict.map(
    tokenize_and_align_labels,
    batched=True,
    remove_columns=dataset_dict["train"].column_names
)
```

```python

########################################
# 5) ENCODE IOB TAGS AS IDS
########################################
# We'll have 3 tags: "O", "B-PHRASE", "I-PHRASE"
tag2id = {"O": 0, "B-PHRASE": 1, "I-PHRASE": 2}
id2tag = {v: k for k, v in tag2id.items()}

def encode_tags(examples):
    encoded_tags = []
    for seq in examples["token_iob"]:
        encoded_seq = [tag2id[tag] if tag in tag2id else 0 for tag in seq]
        encoded_tags.append(encoded_seq)
    examples["token_tags"] = encoded_tags
    return examples

dataset_encoded = dataset_tokenized.map(encode_tags, batched=True)
```

```python

########################################
# 6) CREATE A CUSTOM MULTI-TASK MODEL
########################################
class MultiTaskModel(nn.Module):
    def __init__(self, base_model_name, num_labels_multilabel, num_tags_iob):
        super().__init__()
        self.num_labels_multilabel = num_labels_multilabel
        self.num_tags_iob = num_tags_iob
        
        self.base_model = AutoModel.from_pretrained(base_model_name)
        
        # Classification head (for multi-label: use a linear + sigmoid)
        hidden_size = self.base_model.config.hidden_size
        self.classifier = nn.Linear(hidden_size, num_labels_multilabel)
        
        # Token classification head
        self.token_classifier = nn.Linear(hidden_size, num_tags_iob)
    
    def forward(
        self,
        input_ids,
        attention_mask,
        labels=None,        # multi-label classification
        token_tags=None,    # token classification
    ):
        outputs = self.base_model(
            input_ids=input_ids,
            attention_mask=attention_mask
        )
        # outputs.last_hidden_state shape: [batch_size, seq_len, hidden_size]
        # outputs.pooler_output or DistilBERT doesn't have pooler, so we can take [CLS] token
        # For DistilBERT, let's take the first token's hidden state as the "pooled" representation
        # Or do a mean pooling. We'll do a simple approach:
        
        # 1) multi-label classification
        # DistilBERT doesn't have pooler_output, so let's just use the hidden state of [CLS]-equivalent
        # The first token is typically the "CLS" embedding
        sequence_output = outputs.last_hidden_state
        cls_output = sequence_output[:, 0, :]  # shape [batch, hidden_size]
        
        logits_multilabel = self.classifier(cls_output)  # [batch, num_labels_multilabel]
        # 2) token classification
        token_logits = self.token_classifier(sequence_output)  # [batch, seq_len, num_tags_iob]
        
        loss = None
        if labels is not None:
            # Multi-label: BCEWithLogitsLoss
            bce_loss_fn = nn.BCEWithLogitsLoss()
            loss_multilabel = bce_loss_fn(logits_multilabel, labels.float())
            
            # Token classification: cross-entropy
            # token_tags shape: [batch, seq_len]
            # token_logits shape: [batch, seq_len, num_tags_iob]
            # we can reshape for cross_entropy
            active_loss = attention_mask.view(-1) == 1  # only compute loss for non-padding
            token_logits_reshaped = token_logits.view(-1, self.num_tags_iob)
            token_tags_reshaped = token_tags.view(-1)
            
            # Filter out indices where attention_mask == 0
            active_logits = token_logits_reshaped[active_loss]
            active_labels = token_tags_reshaped[active_loss]
            
            ce_loss_fn = nn.CrossEntropyLoss()
            loss_token = ce_loss_fn(active_logits, active_labels)
            
            loss = loss_multilabel + loss_token
        
        return {
            "loss": loss,
            "logits_multilabel": logits_multilabel,
            "logits_token": token_logits
        }

model = MultiTaskModel(
    base_model_name=model_name,
    num_labels_multilabel=len(label_list),
    num_tags_iob=len(tag2id)
)
```

```python

########################################
# 7) DATA COLLATOR & TRAINER
########################################
# We'll define a data collator that handles token classification padding
# but also we have multi-label "labels" to pass in
class MyDataCollator(DataCollatorForTokenClassification):
    def __call__(self, features):
        # "features" is a list of dict from the dataset
        # We'll separate out the multi-label "labels" and token-level "token_tags"
        labels = [f["labels"] for f in features]
        token_tags = [f["token_tags"] for f in features]
        
        # Convert to torch
        labels_tensor = torch.tensor(labels, dtype=torch.float)
        
        # We rely on DataCollatorForTokenClassification to handle input_ids, attention_mask, etc.
        batch = super().__call__(features)
        
        # Add these extra fields
        batch["labels"] = labels_tensor
        # token_tags is already included as "labels" by default, but we rename it:
        batch["token_tags"] = batch["labels"]
        # Actually, we want both. So let's store them carefully:
        batch["token_tags"] = torch.tensor(token_tags, dtype=torch.long)
        
        # Overwrite "labels" with the multi-label classification
        batch["labels"] = labels_tensor
        
        return batch

data_collator = MyDataCollator(tokenizer, padding=True, max_length=128)

def compute_metrics(eval_preds):
    # We'll only compute something for the multi-label classification for demonstration
    # Typically you'd decode token-level tags and compute token-level metrics if you want
    preds_multilabel = eval_preds.predictions[0]  # "logits_multilabel"
    labels_multilabel = eval_preds.label_ids[0]   # "labels"
    
    # Because we pass them in a dictionary, we might need to check how the Trainer merges them
    # For simplicity, let's do a naive approach:
    preds_sigmoid = 1 / (1 + torch.exp(-torch.tensor(preds_multilabel)))
    preds_binary = (preds_sigmoid >= 0.5).int()
    labels_t = torch.tensor(labels_multilabel)
    
    correct = torch.all(preds_binary == labels_t, dim=1).sum().item()
    accuracy = correct / labels_t.size(0)
    return {"accuracy": accuracy}

# The Hugging Face Trainer expects a single output of "logits" from the model,
# but we have multiple. We'll wrap them or adapt the trainer to parse them.

class CustomTrainer(Trainer):
    def compute_loss(self, model, inputs, return_outputs=False):
        labels = inputs["labels"]        # multi-label
        token_tags = inputs["token_tags"]
        
        outputs = model(
            input_ids=inputs["input_ids"],
            attention_mask=inputs["attention_mask"],
            labels=labels,
            token_tags=token_tags
        )
        loss = outputs["loss"]
        return (loss, outputs) if return_outputs else loss
    
    def prediction_step(
        self, 
        model, 
        inputs, 
        prediction_loss_only, 
        ignore_keys=None
    ):
        # We override to return multi-label logits in a format the metric fn can read
        labels = inputs["labels"]
        token_tags = inputs["token_tags"]
        
        with torch.no_grad():
            outputs = model(
                input_ids=inputs["input_ids"],
                attention_mask=inputs["attention_mask"],
                labels=labels,
                token_tags=token_tags
            )
        loss = outputs["loss"]
        
        # logits for multi-label
        logits_multilabel = outputs["logits_multilabel"].detach().cpu().numpy()
        labels_multilabel = labels.detach().cpu().numpy()
        
        if prediction_loss_only:
            return (loss, None, None)
        
        # We'll return a tuple of (loss, predictions, labels)
        # The Trainer then passes these to compute_metrics
        return (loss, (logits_multilabel, ), (labels_multilabel, ))

training_args = TrainingArguments(
    output_dir="my_mtl_model",
    evaluation_strategy="epoch",
    num_train_epochs=3,
    per_device_train_batch_size=4,  # small for CPU
    per_device_eval_batch_size=4,
    logging_steps=50,
    save_steps=100,
)

trainer = CustomTrainer(
    model=model,
    args=training_args,
    train_dataset=dataset_encoded["train"],
    eval_dataset=dataset_encoded["test"],
    data_collator=data_collator,
    tokenizer=tokenizer,
    compute_metrics=compute_metrics
)

```

```python

########################################
# 8) TRAIN
########################################
trainer.train()
```
